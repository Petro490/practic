```python
class Library:
    def __init__(self):
        # Инициализируем пустой список для хранения названий книг
        self.books = []

    def add_book(self, book_name):
        """Добавляет книгу в список."""
        self.books.append(book_name)
        print(f'Книга "{book_name}" успешно добавлена в библиотеку.')

    def remove_book(self, book_name):
        """Удаляет книгу из списка. Если книги нет — выводит сообщение."""
        if book_name in self.books:
            self.books.remove(book_name)
            print(f'Книга "{book_name}" успешно удалена из библиотеки.')
        else:
            print(f'Ошибка: Книги "{book_name}" нет в списке библиотеки.')

    def display_books(self):
        """Выводит список всех книг. Если библиотека пуста — сообщает об этом."""
        if not self.books:
            print("Библиотека пуста. В ней пока нет ни одной книги.")
        else:
            print("Список книг в библиотеке:")
            for index, book in enumerate(self.books, 1):
                print(f"  {index}. {book}")


# --- Код для проверки работы класса (демонстрация) ---
def test_library():
    print("=== Тестирование класса Library ===")
    
    # 1. Создаем объект библиотеки
    my_library = Library()
    
    # 2. Проверяем отображение книг в пустой библиотеке
    my_library.display_books()
    print("-" * 40)
    
    # 3. Добавляем книги
    my_library.add_book("Изучаем Python")
    my_library.add_book("Чистый код")
    my_library.add_book("Гарри Поттер")
    print("-" * 40)
    
    # 4. Выводим список добавленных книг
    my_library.display_books()
    print("-" * 40)
    
    # 5. Удаляем существующую книгу
    my_library.remove_book("Чистый код")
    
    # 6. Пробуем удалить книгу, которой нет в списке
    my_library.remove_book("Война и мир")
    print("-" * 40)
    
    # 7. Финальный вывод списка книг
    my_library.display_books()

# Запускаем проверку
test_library()
```

```python
class Stack:
    def __init__(self):
        # Инициализируем пустой список для хранения элементов стека
        self.stack = []

    def push(self, action):
        """Добавляет действие (строку) в конец стека."""
        self.stack.append(action)
        print(f'Действие "{action}" добавлено в стек.')

    def pop(self):
        """Удаляет и возвращает последнее добавленное действие. 
        Если стек пуст — сообщает об этом и возвращает None."""
        if self.is_empty():
            print("Ошибка: Стек действий пуст. Нечего отменять!")
            return None
        action = self.stack.pop()
        return action

    def peek(self):
        """Возвращает последнее действие без его удаления. 
        Если стек пуст — сообщает об этом и возвращает None."""
        if self.is_empty():
            print("Предупреждение: Стек действий пуст.")
            return None
        return self.stack[-1]

    def is_empty(self):
        """Возвращает True, если стек пуст, и False в противном случае."""
        return len(self.stack) == 0


# --- Код для проверки работы класса (демонстрация) ---
def test_stack():
    print("=== Тестирование класса Stack (Текстовый редактор) ===")
    
    # 1. Создаем объект стека для истории действий
    editor_history = Stack()
    
    # 2. Проверяем, пуст ли стек в начале
    print(f"Стек пуст? {editor_history.is_empty()}")
    print("-" * 50)
    
    # 3. Пользователь совершает действия в текстовом редакторе
    editor_history.push("Ввод текста: 'Привет'")
    editor_history.push("Выделение текста курсивом")
    editor_history.push("Ввод текста: ' Мир!'")
    print("-" * 50)
    
    # 4. Смотрим на последнее действие без удаления (метод peek)
    last_action = editor_history.peek()
    print(f"Последнее совершенное действие: '{last_action}'")
    print("-" * 50)
    
    # 5. Имитируем функцию отмены действий (Ctrl + Z) с помощью метода pop
    print(f"-> Отмена действия: '{editor_history.pop()}'")
    print(f"-> Отмена действия: '{editor_history.pop()}'")
    print("-" * 50)
    
    # 6. Снова смотрим на последнее оставшееся действие
    print(f"Текущее последнее действие в истории: '{editor_history.peek()}'")
    print("-" * 50)
    
    # 7. Отменяем последнее действие и пробуем сделать отмену ещё раз из пустого стека
    print(f"-> Отмена действия: '{editor_history.pop()}'")
    print(f"Стек пуст? {editor_history.is_empty()}")
    
    # Попытка вызова pop из пустого стека
    editor_history.pop()

# Запускаем проверку
test_stack()
```

```python
class Queue:
    def __init__(self):
        # Инициализируем пустой список для хранения элементов очереди
        self.queue = []

    def enqueue(self, order):
        """Добавляет заказ (строку) в конец очереди."""
        self.queue.append(order)
        print(f'Заказ "{order}" успешно добавлен в очередь.')

    def dequeue(self):
        """Удаляет и возвращает первый заказ из очереди.
        Если очередь пуста — сообщает об этом и возвращает None."""
        if self.is_empty():
            print("Ошибка: Очередь заказов пуста. Нечего обрабатывать!")
            return None
        # Метод pop(0) удаляет и возвращает самый первый (нулевой) элемент списка
        order = self.queue.pop(0)
        return order

    def is_empty(self):
        """Возвращает True, если очередь пуста, и False в противном случае."""
        return len(self.queue) == 0


# --- Код для проверки работы класса (демонстрация) ---
def test_queue():
    print("=== Тестирование класса Queue (Очередь в кафе) ===")
    
    # 1. Создаем объект очереди для заказов кафе
    cafe_orders = Queue()
    
    # 2. Проверяем, пуста ли очередь изначально
    print(f"Очередь пуста? {cafe_orders.is_empty()}")
    print("-" * 50)
    
    # 3. Клиенты делают заказы, добавляем их в конец очереди (enqueue)
    cafe_orders.enqueue("Капучино и круассан")
    cafe_orders.enqueue("Латте на миндальном молоке")
    cafe_orders.enqueue("Эспрессо и чизкейк")
    print("-" * 50)
    
    # 4. Проверяем статус очереди
    print(f"Очередь пуста? {cafe_orders.is_empty()}")
    print("-" * 50)
    
    # 5. Бариста начинает готовить заказы по очереди (FIFO)
    print(f"-> Выдан заказ: '{cafe_orders.dequeue()}'")
    print(f"-> Выдан заказ: '{cafe_orders.dequeue()}'")
    print("-" * 50)
    
    # 6. Добавляем еще один заказ в конец оставшейся очереди
    cafe_orders.enqueue("Зеленый чай и маффин")
    print("-" * 50)
    
    # 7. Обрабатываем все оставшиеся заказы
    print(f"-> Выдан заказ: '{cafe_orders.dequeue()}'")
    print(f"-> Выдан заказ: '{cafe_orders.dequeue()}'")
    print("-" * 50)
    
    # 8. Пробуем вызвать dequeue из пустой очереди
    print(f"Очередь пуста? {cafe_orders.is_empty()}")
    cafe_orders.dequeue()

# Запускаем проверку
test_queue()
```

```python
class HashTable:
    def __init__(self, size=10):
        # Задаем фиксированный размер массива
        self.size = size
        # Создаем массив фиксированного размера, заполненный пустыми списками (метод цепочек)
        self.table = [[] for _ in range(self.size)]

    def _hash_function(self, key):
        """Внутренняя хеш-функция.
        Преобразует ключ в индекс массива в пределах от 0 до (size - 1)."""
        return hash(key) % self.size

    def insert(self, key, value):
        """Добавляет пару ключ-значение в хеш-таблицу.
        Если ключ уже существует, обновляет его значение."""
        index = self._hash_function(key)
        
        # Проверяем, есть ли уже такой ключ в цепочке (списке) по этому индексу
        for pair in self.table[index]:
            if pair[0] == key:
                pair[1] = value  # Если ключ найден, обновляем значение
                print(f'Ключ "{key}" уже существовал. Значение обновлено на "{value}".')
                return
                
        # Если ключа в цепочке не было, добавляем новую пару [ключ, значение]
        self.table[index].append([key, value])
        print(f'Пользователь с ID "{key}" успешно добавлен (Индекс ячейки: {index}).')

    def get(self, key):
        """Возвращает значение по ключу. Если ключ не найден — возвращает None."""
        index = self._hash_function(key)
        
        # Ищем ключ внутри цепочки по полученному индексу
        for pair in self.table[index]:
            if pair[0] == key:
                return pair[1]  # Ключ найден, возвращаем значение
                
        return None  # Ключ не найден

    def remove(self, key):
        """Удаляет пару ключ-значение из хеш-таблицы по ключу."""
        index = self._hash_function(key)
        
        # Ищем пару в цепочке по индексу
        for i, pair in enumerate(self.table[index]):
            if pair[0] == key:
                self.table[index].pop(i)  # Удаляем пару из списка
                print(f'Пользователь с ID "{key}" успешно удален.')
                return True
                
        print(f'Ошибка: Пользователь с ID "{key}" не найден для удаления.')
        return False


# --- Код для проверки работы класса (демонстрация) ---
def test_hash_table():
    print("=== Тестирование класса HashTable ===")
    
    # 1. Создаем хеш-таблицу небольшого размера (например, 5 ячеек), чтобы наглядно увидеть коллизии
    user_db = HashTable(size=5)
    print("-" * 60)
    
    # 2. Добавляем пользователей (метод insert)
    user_db.insert("user_101", {"name": "Алексей", "role": "admin"})
    user_db.insert("user_102", {"name": "Мария", "role": "manager"})
    user_db.insert("user_205", {"name": "Иван", "role": "user"})  # Возможна коллизия по индексу
    print("-" * 60)
    
    # 3. Ищем существующих и несуществующих пользователей (метод get)
    print("Получаем данные user_102:")
    print(f"  Результат: {user_db.get('user_102')}")
    
    print("\nПолучаем данные несуществующего user_999:")
    print(f"  Результат: {user_db.get('user_999')}")
    print("-" * 60)
    
    # 4. Обновляем данные существующего пользователя
    user_db.insert("user_101", {"name": "Алексей", "role": "super_admin"})
    print(f"  Новые данные user_101: {user_db.get('user_101')}")
    print("-" * 60)
    
    # 5. Удаляем пользователя (метод remove)
    user_db.remove("user_102")
    print(f"Проверка присутствия user_102 после удаления: {user_db.get('user_102')}")
    
    # Попытка удалить того, кого нет
    user_db.remove("user_102")

# Запускаем проверку
test_hash_table()
```

```python
def binary_search(sorted_list, target):
    """
    Выполняет бинарный поиск элемента target в отсортированном списке sorted_list.
    Возвращает индекс элемента, если он найден, или -1, если элемент отсутствует.
    """
    # Рекомендация: Убеждаемся, что список отсортирован (для корректной работы алгоритма)
    # Если на вход может прийти неотсортированный список, раскомментируйте строку ниже:
    # sorted_list = sorted(sorted_list)

    # Рекомендация: Инициализируем левый и правый указатели диапазона поиска
    left = 0
    right = len(sorted_list) - 1

    # Цикл продолжается, пока диапазон поиска не схлопнется
    while left <= right:
        # Находим средний индекс текущего диапазона
        mid = (left + right) // 2
        mid_element = sorted_list[mid]

        # Если элемент посередине равен целевому — мы его нашли!
        if mid_element == target:
            return mid
        
        # Рекомендация: Если элемент меньше целевого значения, 
        # отбрасываем левую половину и смещаем левый указатель вправо
        elif mid_element < target:
            left = mid + 1
            
        # Рекомендация: Если элемент больше целевого значения,
        # отбрасываем правую половину и смещаем правый указатель влево
        else:
            right = mid - 1

    # Если цикл завершился, значит, товар в списке не найден
    return -1


# --- Код для проверки работы функции (демонстрация) ---
def test_binary_search():
    print("=== Тестирование функции binary_search ===")
    
    # Отсортированный список уникальных ID товаров в магазине
    product_ids = [1001, 1005, 1012, 1025, 1042, 1050, 1088, 1100]
    print(f"Список товаров в базе: {product_ids}")
    print("-" * 50)
    
    # Тест 1: Ищем товар, который есть в середине списка
    target_1 = 1042
    index_1 = binary_search(product_ids, target_1)
    print(f"Поиск товара с ID {target_1}:")
    print(f"  Результат: Индекс {index_1} (Элемент на этом месте: {product_ids[index_1]})")
    print("-" * 50)
    
    # Тест 2: Ищем товар, который находится в самом начале списка (крайний случай)
    target_2 = 1001
    index_2 = binary_search(product_ids, target_2)
    print(f"Поиск товара с ID {target_2}:")
    print(f"  Результат: Индекс {index_2} (Элемент на этом месте: {product_ids[index_2]})")
    print("-" * 50)
    
    # Тест 3: Ищем товар, которого вообще нет в магазине
    target_3 = 1060
    index_3 = binary_search(product_ids, target_3)
    print(f"Поиск товара с ID {target_3}:")
    print(f"  Результат: Код {index_3} (Товар не найден)")
    print("-" * 50)

# Запускаем проверку
test_binary_search()
```

```python
# Альтернативный вариант работы с Hash
import time

# --- 1. Ваша ручная HashTable (для сравнения) ---
class HashTable:
    def __init__(self, size=1000): # Дадим ей хороший размер, чтобы цепочки были короче
        self.size = size
        self.table = [[] for _ in range(self.size)]

    def _hash_function(self, key):
        return hash(key) % self.size

    def insert(self, key, value):
        index = self._hash_function(key)
        for pair in self.table[index]:
            if pair[0] == key:
                pair[1] = value
                return
        self.table[index].append([key, value])

    def get(self, key):
        index = self._hash_function(key)
        for pair in self.table[index]:
            if pair[0] == key:
                return pair[1]
        return None

# --- 2. Подготовка тестовых данных ---
N = 50000  # Количество операций
keys = [f"user_{i}" for i in range(N)]
values = [f"data_{i}" for i in range(N)]

print(f"Запуск бенчмарка на {N} элементов...\n")

# --- 3. Тестируем РУЧНУЮ HashTable ---
custom_table = HashTable(size=5000)

# Замер вставки
start_time = time.perf_counter()
for k, v in zip(keys, values):
    custom_table.insert(k, v)
custom_insert_time = (time.perf_counter() - start_time) * 1000

# Замер поиска
start_time = time.perf_counter()
for k in keys:
    custom_table.get(k)
custom_get_time = (time.perf_counter() - start_time) * 1000


# --- 4. Тестируем ВСТРОЕННЫЙ dict ---
native_dict = {}

# Замер вставки
start_time = time.perf_counter()
for k, v in zip(keys, values):
    native_dict[k] = v
native_insert_time = (time.perf_counter() - start_time) * 1000

# Замер поиска
start_time = time.perf_counter()
for k in keys:
    _ = native_dict[k]
native_get_time = (time.perf_counter() - start_time) * 1000


# --- 5. Вывод результатов ---
print("📊 РЕЗУЛЬТАТЫ (в миллисекундах):")
print("-" * 50)
print(f"Ручная HashTable | Вставка: {custom_insert_time:.2f} мс | Поиск: {custom_get_time:.2f} мс")
print(f"Встроенный dict  | Вставка: {native_insert_time:.2f} мс | Поиск: {native_get_time:.2f} мс")
print("-" * 50)
print(f"🚀 Встроенный dict быстрее вставки в {custom_insert_time / native_insert_time:.1f} раз!")
print(f"🚀 Встроенный dict быстрее поиска в {custom_get_time / native_get_time:.1f} раз!")
```
