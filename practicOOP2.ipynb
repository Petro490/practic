```python
# задание 4.
import functools

class Millimeter:
    def __init__(self, value):
        if isinstance(value, (int, float)):
            self._value = float(value)
        else:
            raise TypeError("Value must be an integer or float for Millimeter.")
        self.label = "Millimeter"
        self.ratio = 1.0

    def as_millimeters(self):
        """Возвращает значение объекта в миллиметрах."""
        return self._value * self.ratio

    def __repr__(self):
        """Возвращает строковое представление объекта."""
        return f"{self.label}({self._value})"

    def _convert_to_same_type(self, result_in_mm):
        """Вспомогательный метод для конвертации миллиметров обратно в тип self."""
        value_in_current_unit = result_in_mm / self.ratio
        return type(self)(value_in_current_unit)

    # --- Арифметические операции ---
    def __add__(self, other):
        if isinstance(other, Millimeter):
            self_in_mm = self.as_millimeters()
            other_in_mm = other.as_millimeters()
        elif isinstance(other, (int, float)):
            other_in_mm = float(other)
            self_in_mm = self.as_millimeters()
        else:
            return NotImplemented
        result_in_mm = self_in_mm + other_in_mm
        return self._convert_to_same_type(result_in_mm)

    def __radd__(self, other):
        if isinstance(other, (int, float)):
            other_in_mm = float(other)
            self_in_mm = self.as_millimeters()
            result_in_mm = other_in_mm + self_in_mm
            return self._convert_to_same_type(result_in_mm)
        else:
            return NotImplemented

    def __sub__(self, other):
        if isinstance(other, Millimeter):
            self_in_mm = self.as_millimeters()
            other_in_mm = other.as_millimeters()
        elif isinstance(other, (int, float)):
            other_in_mm = float(other)
            self_in_mm = self.as_millimeters()
        else:
            return NotImplemented
        result_in_mm = self_in_mm - other_in_mm
        return self._convert_to_same_type(result_in_mm)

    def __rsub__(self, other):
        if isinstance(other, (int, float)):
            other_in_mm = float(other)
            self_in_mm = self.as_millimeters()
            result_in_mm = other_in_mm - self_in_mm
            return self._convert_to_same_type(result_in_mm)
        else:
            return NotImplemented

    def __mul__(self, factor):
        if isinstance(factor, (int, float)):
            result_in_mm = self.as_millimeters() * float(factor)
            return self._convert_to_same_type(result_in_mm)
        else:
            return NotImplemented

    def __rmul__(self, factor):
        return self.__mul__(factor)

    def __truediv__(self, factor):
        if isinstance(factor, (int, float)):
            if factor == 0:
                raise ZeroDivisionError("division by zero")
            result_in_mm = self.as_millimeters() / float(factor)
            return self._convert_to_same_type(result_in_mm)
        else:
            return NotImplemented

    # --- Методы сравнения и хэширования ---
    @functools.total_ordering
    def __eq__(self, other):
        if not isinstance(other, Millimeter):
            return NotImplemented
        return abs(self.as_millimeters() - other.as_millimeters()) < 1e-9

    def __lt__(self, other):
        if not isinstance(other, Millimeter):
            return NotImplemented
        return self.as_millimeters() < other.as_millimeters()

    def __le__(self, other):
        if not isinstance(other, Millimeter):
            return NotImplemented
        return self.as_millimeters() <= other.as_millimeters()

    def __hash__(self):
        return hash(self.as_millimeters())

    # --- Новые методы для преобразования ---
    def __int__(self):
        """Возвращает значение объекта в миллиметрах как целое число."""
        # Округляем до ближайшего целого, так как int() обрезает дробную часть.
        # Выбор метода округления (round, floor, ceil) зависит от требований.
        # round() - стандартное округление.
        return int(round(self.as_millimeters()))

    def __float__(self):
        """Возвращает значение объекта в миллиметрах как число с плавающей точкой."""
        return float(self.as_millimeters())

# --- Дочерние классы ---
# Они автоматически наследуют методы __int__ и __float__.

class Centimeter(Millimeter):
    def __init__(self, value):
        self.label = "Centimeter"
        self.ratio = 10.0
        if isinstance(value, (int, float)):
            self._value = float(value)
        elif isinstance(value, Millimeter):
            self._value = value.as_millimeters() / self.ratio
        else:
            raise TypeError("Unsupported type for Centimeter initialization.")
    def __repr__(self):
        return f"{self.label}({self._value})"

class Meter(Millimeter):
    def __init__(self, value):
        self.label = "Meter"
        self.ratio = 1000.0
        if isinstance(value, (int, float)):
            self._value = float(value)
        elif isinstance(value, Millimeter):
            self._value = value.as_millimeters() / self.ratio
        else:
            raise TypeError("Unsupported type for Meter initialization.")
    def __repr__(self):
        return f"{self.label}({self._value})"

class Inch(Millimeter):
    def __init__(self, value):
        self.label = "Inch"
        self.ratio = 25.4
        if isinstance(value, (int, float)):
            self._value = float(value)
        elif isinstance(value, Millimeter):
            self._value = value.as_millimeters() / self.ratio
        else:
            raise TypeError("Unsupported type for Inch initialization.")
    def __repr__(self):
        return f"{self.label}({self._value})"


# Создаем объекты
m1 = Meter(2)        # 2000 мм
cm1 = Centimeter(50)  # 500 мм
inch1 = Inch(10)     # 254 мм
mm1 = Millimeter(500.7) # 500.7 мм

print("--- Преобразование в float ---")
float_m1 = float(m1)
print(f"float({m1!r}) = {float_m1} (тип: {type(float_m1)})")

float_inch1 = float(inch1)
print(f"float({inch1!r}) = {float_inch1} (тип: {type(float_inch1)})")

print("\n--- Преобразование в int ---")
int_m1 = int(m1) # 2000 мм -> 2000
print(f"int({m1!r}) = {int_m1} (тип: {type(int_m1)})")

int_inch1 = int(inch1) # 254 мм -> 254
print(f"int({inch1!r}) = {int_inch1} (тип: {type(int_inch1)})")

int_mm1 = int(mm1) # 500.7 мм -> 501 (из-за round)
print(f"int({mm1!r}) = {int_mm1} (тип: {type(int_mm1)})")

# Пример использования в вычислениях
total_mm = m1 + cm1 # 2000 мм + 500 мм = 2500 мм
print(f"\nСумма {m1!r} и {cm1!r} = {total_mm!r}")
print(f"Преобразование суммы в int: {int(total_mm)}")
print(f"Преобразование суммы в float: {float(total_mm)}")
```
